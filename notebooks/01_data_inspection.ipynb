Investigate missing fast-charging specifications

```python
df.groupby(
    "fast_charging_available",
    dropna=False
)["fast_charging"].agg(
    total_rows = "size",
    missing_values = lambda s: s.isna().sum(),
    available_values = lambda s: s.notna().sum()
    
)
```

```python
# Find repeated combinations
model_duplicates = df.duplicated(
    subset=["brand_name","model"],
    keep = False
)

df.loc[
    model_duplicates,
    ["brand_name","model","price"]
].sort_values(["brand_name","model"])
```

```python
# Count the repeated combinations
repeated_rows = df.duplicated(
    subset=["brand_name","model"],
    keep=False
)

print("Rows belonging to repeated  combination:",repeated_rows.sum())

print(
    "distinct repeated brand/model combination: ",
    df.loc[repeated_rows,["brand_name","model"]].drop_duplicates().shape[0]
)
```

```python
# Investigate differences in price

df.loc[repeated_rows].groupby(
    ["brand_name","model"]
).agg(
    records = ("price","size"),
    unique_prices = ("price","nunique"),
    minimum_prices = ("price","min"),
    maximum_prices = ("price","max")
).sort_values("records",ascending=False)
```
